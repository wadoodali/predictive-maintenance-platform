# Data Preprocessing

This notebook prepares the AI4I 2020 dataset for machine learning.
We clean the data, select the relevant features, encode categorical data,
and split the data into training and test sets.

```python
import pandas as pd

df = pd.read_csv("../data/raw/ai4i2020.csv")
df.head()
```

```python
X = df[['Type', 'Air temperature [K]', 'Process temperature [K]',
        'Rotational speed [rpm]', 'Torque [Nm]', 'Tool wear [min]']]

y = df['Machine failure']
```

```python
# One-hot encoding to convert 'Type' into numerical feature
X = pd.get_dummies(X, columns=['Type'], dtype=int)
```

```python
# Splitting into Train/Test
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)
```

```python
# Feature Scaling
from sklearn.preprocessing import StandardScaler

scaler  = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```
